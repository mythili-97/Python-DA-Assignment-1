```python
#Numpy Array Operations:
import numpy as np
```

```python
#1D NumPy array:
temperatures_w1=[22.5,25.3,20.8,23.4,26.1,24.8,21.9]
arr=np.array(temperatures_w1)
arr
```

```python
#Inspection and properties:
arr.dtype
print("Data type:",arr.dtype)
arr.size
print("Number of elements:",arr.size)
arr.shape
print("Shape:",arr.shape)
```

```python
#Array operations:
Fahrenheit=(arr*9/5)+32
print(Fahrenheit)
```

```python
np.max(arr)
print("Maximum Temp:",np.max(arr))
np.min(arr)
print("Minimum Temp:",np.min(arr))
np.mean(arr)
print("Mean Temp:",np.mean(arr))
```

```python
#array slicing and Indexing:
arr[0:3]
print("Temperature for first three days:",arr[0:3])
arr[-2:]
print("Temperature for last two days:",arr[-2:])
arr[2:5]
print("Middle Three days:",arr[2:5])
```

```python
#2D Array:
Temperatures=[
              [22.5,25.3,20.8,23.4,26.1,24.8,21.9],
              [19.2,22.5,21.3,24.0,23.5,22.8,20.1]
]
arr_2d=np.array(Temperatures)
arr_2d
```

```python
#Inspecting and Slicing 2D Array:
arr_2d.shape
print("Shape:",arr_2d.shape)
arr_2d.dtype
print("Data Type:",arr_2d.dtype)
arr_2d.size
print("Size:",arr_2d.size)

arr_2d[0:5,0:5]
print("Temperature for each weekdays:",arr_2d[0:5,0:5])
arr_2d[-2:,-2:]
print("Temperature for each weekends:",arr_2d[0:5,0:5])
```

```python
#Pandas Series:
import pandas as pd
```

```python
marks=[95,92,89,85,80]
ser1=pd.Series(marks,index=['Rank1','Rank2','Rank3','Rank4','Rank5'],name="Marks")
ser1
```

```python
#Indexing and Slicing:
ser1.iloc[0]
print("Mark of first student:",ser1.iloc[0])
ser1.loc['Rank1':'Rank3']
print("Mark of top three ranks:",ser1.loc['Rank1':'Rank3'])
ser1.iloc[2]
print("Mark of 3rd rank student:",ser1.iloc[2])
#Boolean Mask:
ser1[ser1 > 90]
```

```python
#Manipulating Series:
ser1.loc["Rank1"]=100
ser1
```

```python
ser1.drop('Rank5',inplace=True)
print(ser1)
```

```python
Cgpa=ser1/10
print(Cgpa)
```
